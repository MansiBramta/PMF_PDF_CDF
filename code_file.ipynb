```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Create values for weight
x = np.linspace(30, 90, 1000)

# Assume weights follow a normal distribution
# Mean = 60 kg, Standard Deviation = 10 kg
mu = 60
sigma = 10

# Calculate the PDF
y = norm.pdf(x, loc=mu, scale=sigma)

# Create the PDF curve
plt.plot(x, y, linewidth=2)

# Select the range between 50 and 60
x_fill = np.linspace(50, 60, 1000)
y_fill = norm.pdf(x_fill, loc=mu, scale=sigma)

# Shade the area representing P(50 < X < 60)
plt.fill_between(x_fill, y_fill, alpha=0.4)

# Add labels
plt.xlabel("Weight (kg)")
plt.ylabel("Probability Density")
plt.title("PDF")

# Add vertical lines at 50 and 60
plt.axvline(50, linestyle="--")
plt.axvline(60, linestyle="--")



plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Possible values of X
x = np.array([0, 1, 2])

# PMF
pmf = np.array([1/4, 2/4, 1/4])

# Cumulative probabilities
cdf = np.cumsum(pmf)

# Plot CDF
plt.step(x, cdf, where="post", linewidth=2)

# Mark the CDF values
plt.scatter(x, cdf, s=50)

plt.xlabel("Number of Heads")
plt.ylabel("Cumulative Probability")
plt.title("CDF of Number of Heads in Two Coin Tosses")

plt.xticks(x)
plt.yticks([0, 0.25, 0.5, 0.75, 1])

plt.ylim(0, 1.1)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Mean and standard deviation
mu = 60
sigma = 10

# Values of X
x = np.linspace(30, 90, 1000)

# PDF
pdf = norm.pdf(x, loc=mu, scale=sigma)

# CDF
cdf = norm.cdf(x, loc=mu, scale=sigma)

# Plot CDF
plt.plot(x, cdf, linewidth=2)

plt.xlabel("Weight (kg)")
plt.ylabel("Cumulative Probability")
plt.title("CDF of Student Weight")

plt.ylim(0, 1.05)

plt.show()
```
